```python
import pandas as pd

# 1. Dataset ko read karna
df = pd.read_csv('data.csv')
print("--- Original Messy Data ---")
print(df.head(25))

# 2. Missing Values handling (Calories ke empty cells ko median se fill karna)
median_calories = df['Calories'].median()
df['Calories'].fillna(median_calories, inplace=True)

# 3. Galat Date Format ko sahi karna aur missing dates wale row ko drop karna
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df.dropna(subset=['Date'], inplace=True)

# 4. Outliers/Wrong rows filter karna (Duration 450 ko thik karke 45 karna)
df.loc[df['Duration'] == 450, 'Duration'] = 45

# 5. Ek naya custom column create karna (Total Pulse = Pulse + Maxpulse)
df['Total_Pulse'] = df['Pulse'] + df['Maxpulse']

print("\n--- Cleaned & Filtered Data ---")
print(df.head(25))
```
